# NarraReel on Colab: your voice, free GPU

Turn a script into a finished video narrated in **your own cloned voice**. Choose **Runtime > Change runtime type > GPU** first.

Only clone a voice that is yours, or one you have explicit permission to use. Output carries Resemble AI's inaudible Perth watermark.

```python
#@title 1. Install (about 3 minutes)
!pip install -q "narrareel[clone,align] @ git+https://github.com/SanTiwari07/NarraReel.git"
!narrareel doctor
```

```python
#@title 2. Your script
script = """Wake up early.
Grind hard every single day.

Success is waiting for you at the end of the road.""" #@param {type:"string"}
open("script.txt", "w", encoding="utf-8").write(script)
```

```python
#@title 3. Upload a recording of your voice (15-30 seconds, read it the way you want the video to sound)
from google.colab import files
uploaded = files.upload()
recording = next(iter(uploaded))
print("Using", recording)
```

```python
#@title 4. Save the voice
voice_name = "My voice" #@param {type:"string"}
consent = False #@param {type:"boolean"}
assert consent, "Tick the box to confirm this is your voice, or you have permission to use it."
!narrareel clone "{recording}" --name "{voice_name}" --i-have-consent
```

```python
#@title 5. Make the video
import re
voice_id = re.sub(r"[^a-z0-9]+", "-", voice_name.lower()).strip("-")
energy = 0.8 #@param {type:"slider", min:0, max:1, step:0.05}
length = "30s" #@param {type:"string"}
style = "bold-pop" #@param ["bold-pop", "karaoke", "boxed", "one-word", "minimal", "neon", "keyword-emphasis", "subtitle-classic"]
format = "reels" #@param ["reels", "youtube", "square", "portrait"]
!narrareel make script.txt --voice chatterbox:{voice_id} --energy {energy} --target {length} --style {style} --format {format} -o /content/video.mp4
```

```python
#@title 6. Download
from google.colab import files
files.download("/content/video.mp4")
```
